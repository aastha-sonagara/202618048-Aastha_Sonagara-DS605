### Observation 1 – Price Distribution

From the price distribution graph, we can observe that most books are priced between £15 and £55.This shows that the website mainly contains books in the medium price range rather than very cheap or very expensive books.

### Observation 2 – Rating Distribution

The rating distribution shows that the books are fairly evenly distributed across all five ratings. 3-star books appear slightly more, while 4-star books are slightly fewer compared to the other ratings. This indicates that the dataset contains a balanced mix of books with different quality ratings.

### Observation 3 – Category Representation

According to the category frequency graph, Sequential Art has the highest number of books in the dataset, followed by Nonfiction and Default. Categories such as Science, Politics, Art, Travel, and Health contain only a few books. This means the dataset is not evenly distributed among all categories.

### Observation - 4 Best Value Books

The value_score feature combines rating and affordability. Books with high ratings and lower prices receive higher value scores. These books can be considered better value because they offer good ratings without being expensive.

### Observation 5 – Category-wise Pricing

The average price by category graph shows that Historical Fiction has the highest average book price, followed by Politics, Childrens, and Health. On the other hand, Young Adult and Spirituality have the lowest average prices. This suggests that book prices vary noticeably across different categories.

### Observation 6 – Relationship Between Price and Rating

The Price vs Rating scatter plot does not show any clear upward or downward trend. Books with both low and high ratings are available across almost every price range. Therefore, price does not appear to have a strong relationship with rating. Spending more money does not necessarily mean the book will have a higher rating.

### Observation 7 – Book Descriptions and Word Cloud

The word cloud highlights words such as life, one, world, time, book, story, love, family, friend, and young as the most common words in the book descriptions. This suggests that many books focus on everyday life, relationships, personal experiences, and fictional storytelling. These themes seem to appear across multiple categories.

### Conclusion

This dataset provides useful information about book categories, prices, ratings, stock availability, and descriptions. It helps compare different genres and understand general pricing patterns. However, price alone is not a good indicator of book quality, as ratings are spread across almost every price range.
