```python
import pandas as pd
import numpy as np
import re
```

```python
# ==========================================================
# Load the scraped dataset
# ==========================================================

df = pd.read_csv("../data/books.csv")

print("Dataset Loaded Successfully")

df.head()
```

```python
# ==========================================================
# Display dataset information
# ==========================================================

print("Shape of Dataset :", df.shape)

print("\nColumn Names\n")
print(df.columns)

print("\nDataset Information\n")
df.info()
```

```python
# ==========================================================
# Statistical Summary
# ==========================================================

df.describe(include="all")
```

```python
# ==========================================================
# Check Missing Values
# ==========================================================

missing_values = df.isnull().sum()

print(missing_values)
```

```python
# ==========================================================
# Percentage Missing Values
# ==========================================================

missing_percentage = (df.isnull().sum()/len(df))*100

missing_percentage
```

```python
# ==========================================================
# Check Duplicate Records
# ==========================================================

duplicates = df.duplicated().sum()

print("Duplicate Rows :", duplicates)
```

```python
# ==========================================================
# Remove Duplicate Rows
# ==========================================================

df.drop_duplicates(inplace=True)

print("New Shape :", df.shape)
```

```python
# ==========================================================
# Check Duplicate UPC
# ==========================================================

duplicate_upc = df["UPC"].duplicated().sum()

print("Duplicate UPC :", duplicate_upc)
```

```python
# ==========================================================
# Remove Pound Symbol
# ==========================================================

df["Price"] = df["Price"].str.replace("£","",regex=False)

df["Price"] = df["Price"].astype(float)

df.head()
```

```python
# ==========================================================
# Convert Rating into Numeric Values
# ==========================================================

rating_map = {

    "One":1,

    "Two":2,

    "Three":3,

    "Four":4,

    "Five":5

}

df["Rating"] = df["Rating"].map(rating_map)

df.head()
```

```python
# ==========================================================
# Clean Availability Column
# ==========================================================

df["Availability"] = df["Availability"].str.strip()

df.head()
```

```python
# ==========================================================
# Convert Reviews into Integer
# ==========================================================

df["Number of Reviews"] = df["Number of Reviews"].astype(int)

df.info()
```

```python
# ==========================================================
# Fill Missing Description
# ==========================================================

df["Product Description"] = df["Product Description"].fillna("No Description Available")
```

```python
# ==========================================================
# Verify Data Types
# ==========================================================

df.dtypes
```

```python
# ==========================================================
# Final Missing Value Check
# ==========================================================

df.isnull().sum()
```

```python
# ==========================================================
# Final Dataset Information
# ==========================================================

df.info()
```

```python
# ==========================================================
# Save Cleaned Dataset
# ==========================================================

df.to_csv("../data/books_cleaned.csv",index=False)

print("Cleaned Dataset Saved Successfully")
```

```python
# ==========================================================
# Display Clean Dataset
# ==========================================================

df.head(10)
```

```python
# ==========================================================
# Extract Available Stock Count from Availability Column
# ==========================================================

import re

def extract_stock(text):
    """
    Extract the number of books available in stock.
    Example:
        'In stock (22 available)' -> 22
        'In stock' -> 0
    """

    if pd.isna(text):
        return 0

    match = re.search(r'(\d+)', text)

    if match:
        return int(match.group(1))
    else:
        return 0


df["Available Stock"] = df["Availability"].apply(extract_stock)

df[["Availability", "Available Stock"]].head()
```

```python
# ==========================================================
# Feature 1 : Description Word Count
# ==========================================================

df["description_word_count"] = (
    df["Product Description"]
    .fillna("")
    .apply(lambda x: len(str(x).split()))
)

df[["Product Description", "description_word_count"]].head()
```

```python
# ==========================================================
# Feature 2 : Price Band
# ==========================================================

df["price_band"] = pd.cut(
    df["Price"],
    bins=[0, 20, 35, 50, float("inf")],
    labels=["Budget", "Economy", "Premium", "Luxury"]
)

df[["Price", "price_band"]].head()
```

```python
# ==========================================================
# Feature 3 : Affordability Score
# ==========================================================

max_price = df["Price"].max()

df["affordability_score"] = (
    (max_price - df["Price"]) / max_price
).round(3)

df[["Price", "affordability_score"]].head()
```

```python
# ==========================================================
# Feature 4 : Value Score
# ==========================================================

df["value_score"] = (
    df["Rating"] * df["affordability_score"]
).round(2)

df[["Rating", "Price", "value_score"]].head()
```

```python
# ==========================================================
# Feature 5 : Recommended
# ==========================================================

df["recommended"] = np.where(
    (df["Rating"] >= 4) &
    (df["Price"] <= 35),
    "Yes",
    "No"
)

df[["Rating", "Price", "recommended"]].head()
```

```python
# ==========================================================
# Display Newly Created Features
# ==========================================================

new_features = [

    "Available Stock",

    "description_word_count",

    "price_band",

    "affordability_score",

    "value_score",

    "recommended"

]

df[new_features].head(10)
```

```python
# ==========================================================
# Final Dataset Summary
# ==========================================================

print("Final Shape :", df.shape)

print("\nColumns\n")

print(df.columns)

print("\nData Types\n")

print(df.dtypes)
```

```python
# ==========================================================
# Save Final Cleaned Dataset
# ==========================================================

output_path = "../data/books_cleaned.csv"

df.to_csv(output_path, index=False)

print("===========================================")
print("Dataset Saved Successfully")
print("Location :", output_path)
print("===========================================")
```
