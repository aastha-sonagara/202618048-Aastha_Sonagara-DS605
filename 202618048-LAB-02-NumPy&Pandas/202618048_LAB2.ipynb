##### Task 2 - Vectorized Arithmetic and Linear Algebra

● Create two matrices. Perform addition, element-wise multiplication, and matrix multiplication using @ or np.matmul().
● For a square matrix, find its transpose and determinant.
● If invertible, find the inverse and verify A @ A^-1 using np.allclose().
● Use vectorized NumPy operations; avoid explicit Python loops.

```python
# Create the first matrix A.
# This is a 2 x 2 matrix, meaning it has 2 rows and 2 columns.

A = np.array([
    [1, 2],
    [3, 4]
])

# Create the second matrix B.
# This is also a 2 x 2 matrix.

B = np.array([
    [5, 6],
    [7, 8]
])

# Display both matrices.
print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)
```

```python
# Add matrix A and matrix B.
# The corresponding elements of both matrices are added.

addition = A + B

# Display the result.
print("A + B:")
print(addition)
```

```python
# Perform element-wise multiplication.
# Each element of A is multiplied by the element
# at the same position in B.

element_multiplication = A * B

# Display the result.
print("Element-wise multiplication:")
print(element_multiplication)
```

```python
# Perform matrix multiplication using @.
# The @ operator is used specifically for matrix multiplication.

matrix_multiplication = A @ B

# Display the result.
print("Matrix multiplication:")
print(matrix_multiplication)
```

```python
# Find the transpose of matrix A.
# Transpose changes rows into columns
# and columns into rows.

transpose_A = A.T

# Display the transpose.
print("Transpose of A:")
print(transpose_A)
```

```python
# Find the determinant of matrix A.
# np.linalg.det() is used to calculate the determinant.

determinant_A = np.linalg.det(A)

# Display the determinant.
print("Determinant of A:", determinant_A)
```

```python
# Find the inverse of matrix A.
# np.linalg.inv() calculates the inverse of a matrix.

inverse_A = np.linalg.inv(A)

# Display the inverse.
print("Inverse of A:")
print(inverse_A)
```

```python
# Multiply A by its inverse.
# The result should be the identity matrix.

result = A @ inverse_A

# Display the result.
print("A @ A^-1:")
print(result)

# Check whether the result is approximately
# equal to the identity matrix.
# np.allclose() is used because floating-point calculations
# may produce very small decimal errors.

print("Is A @ A^-1 equal to identity matrix?")
print(np.allclose(result, np.eye(2)))
```

```python
# Display all the results together.

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)

print("\nAddition:")
print(A + B)

print("\nElement-wise multiplication:")
print(A * B)

print("\nMatrix multiplication:")
print(A @ B)

print("\nTranspose of A:")
print(A.T)

print("\nDeterminant of A:")
print(np.linalg.det(A))

print("\nInverse of A:")
print(np.linalg.inv(A))

print("\nVerification using np.allclose():")
print(np.allclose(A @ np.linalg.inv(A), np.eye(2)))
```

```python
# Import NumPy for generating random values
# and performing statistical calculations.

import numpy as np

# Import Matplotlib's pyplot.
# It is used to create and display the histogram.

import matplotlib.pyplot as plt
```

```python
# Choose the mean of the normal distribution.
# We are choosing mean = 50.

mean = 50

# Choose the standard deviation of the normal distribution.
# We are choosing standard deviation = 10.

std = 10

# Display the chosen values.

print("Chosen Mean:", mean)
print("Chosen Standard Deviation:", std)
```

```python
# Set the random seed so that we get the same random values
# every time we run the code.

np.random.seed(42)

# Generate 1,000 random values from a normal distribution.
#
# np.random.normal() takes:
# mean -> the center of the distribution
# std  -> the standard deviation
# 1000 -> the number of values we want

data = np.random.normal(mean, std, 1000)

# Display the first 10 values.
# We display only the first 10 so that the notebook
# does not print all 1,000 values.

print("First 10 values:")
print(data[:10])

# Display the total number of values.

print("Number of values:", len(data))
```

```python
# Calculate the sample standard deviation.
#
# ddof=1 is used because we are calculating the
# standard deviation of a sample.
#
# ddof=1 uses (n-1) in the calculation.

sample_std = np.std(data, ddof=1)

# Display the sample standard deviation.

print("Sample Standard Deviation:", sample_std)
```

```python
# Display the chosen values and the values calculated
# from our generated sample.

print("Chosen Mean:", mean)
print("Sample Mean:", sample_mean)

print("\nChosen Standard Deviation:", std)
print("Sample Standard Deviation:", sample_std)
```

##### Task 5 - Filtering and Querying


Use Boolean indexing and/or query() to answer:
● Male passengers older than 50: how many?
● Female first-class passengers: how many, and what percentage survived?
● Age 20-40, Fare above overall median, and survived: how many?
● Travelling alone (SibSp=0, Parch=0), age < 30, and did not survive: how many?
● Embarked='S', Pclass 2 or 3, and Fare above Southampton median: how many?

```python
# Display the first 5 rows of the Titanic dataset.
# This is just to make sure that our DataFrame "df"
# is loaded and ready to use.

print(df.head())
```

```python
# Select passengers who satisfy BOTH conditions:
#
# Condition 1:
# Sex == "male"
# → passenger must be male
#
# Condition 2:
# Age > 50
# → passenger must be older than 50
#
# The & symbol means AND.
#
# Each condition is written inside brackets because
# Pandas requires brackets around individual conditions.

male_older_50 = df[
    (df["Sex"] == "male") &
    (df["Age"] > 50)
]

# Count the number of passengers satisfying both conditions.
# len() gives the number of rows in the filtered DataFrame.

count_male_older_50 = len(male_older_50)

# Display the result.

print("Number of male passengers older than 50:", count_male_older_50)
```

```python
# Select passengers who satisfy BOTH conditions:
#
# Sex == "female"
# → passenger is female
#
# Pclass == 1
# → passenger travelled in first class

female_first_class = df[
    (df["Sex"] == "female") &
    (df["Pclass"] == 1)
]

# Count the number of female first-class passengers.

count_female_first_class = len(female_first_class)

# Display the count.

print("Number of female first-class passengers:",
      count_female_first_class)
```

```python
# In the Survived column:
#
# 1 = survived
# 0 = did not survive
#
# Therefore, calculating the mean of the Survived column
# gives the survival rate because the values are 0 and 1.
#
# We multiply by 100 to convert the rate into a percentage.

survival_percentage = female_first_class["Survived"].mean() * 100

# Display the survival percentage.

print("Percentage of female first-class passengers who survived:",
      survival_percentage)
```

```python
# Find the median Fare for the entire Titanic dataset.
#
# median() returns the middle value after arranging
# all the Fare values in order.

overall_median_fare = df["Fare"].median()

# Display the median Fare.

print("Overall median Fare:", overall_median_fare)
```

```python
# Select passengers who satisfy ALL three conditions:
#
# 1. Age >= 20
#    → age is 20 or older
#
# 2. Age <= 40
#    → age is 40 or younger
#
# Together, these two conditions mean:
# Age is between 20 and 40.
#
# 3. Fare > overall_median_fare
#    → Fare is greater than the overall median Fare
#
# 4. Survived == 1
#    → passenger survived
#
# The & symbol means AND.

age_20_40_high_fare_survived = df[
    (df["Age"] >= 20) &
    (df["Age"] <= 40) &
    (df["Fare"] > overall_median_fare) &
    (df["Survived"] == 1)
]

# Count the number of passengers satisfying all conditions.

count_age_20_40 = len(age_20_40_high_fare_survived)

# Display the result.

print("Number of passengers:", count_age_20_40)
```

```python
# Select passengers satisfying ALL four conditions:
#
# SibSp == 0
# → no siblings or spouse
#
# Parch == 0
# → no parents or children
#
# Age < 30
# → passenger is younger than 30
#
# Survived == 0
# → passenger did not survive

alone_young_not_survived = df[
    (df["SibSp"] == 0) &
    (df["Parch"] == 0) &
    (df["Age"] < 30) &
    (df["Survived"] == 0)
]

# Count the number of passengers satisfying all conditions.

count_alone_young_not_survived = len(alone_young_not_survived)

# Display the result.

print("Number of passengers:", count_alone_young_not_survived)
```

```python
# First, select only passengers who embarked from Southampton.
#
# Embarked == "S"
# → passenger boarded the ship at Southampton.

southampton_passengers = df[df["Embarked"] == "S"]

# Calculate the median Fare for only these passengers.

southampton_median_fare = southampton_passengers["Fare"].median()

# Display the Southampton median Fare.

print("Southampton median Fare:", southampton_median_fare)
```

```python
# Select passengers satisfying ALL conditions:
#
# 1. Embarked == "S"
#    → passenger embarked at Southampton
#
# 2. Pclass == 2 OR Pclass == 3
#    → passenger was in second or third class
#
# 3. Fare > southampton_median_fare
#    → Fare is greater than the median Fare
#       of Southampton passengers
#
# The | symbol means OR.
#
# The & symbol means AND.

south_pclass_high_fare = df[
    (df["Embarked"] == "S") &
    (
        (df["Pclass"] == 2) |
        (df["Pclass"] == 3)
    ) &
    (df["Fare"] > southampton_median_fare)
]

# Count the passengers satisfying all conditions.

count_south_pclass_high_fare = len(south_pclass_high_fare)

# Display the result.

print("Number of passengers:", count_south_pclass_high_fare)
```

```python
# Display all the answers from Task 5 together.

print("TASK 5 RESULTS")
print("----------------------------")

# 1. Male passengers older than 50

print("1. Male passengers older than 50:",
      count_male_older_50)

# 2. Female first-class passengers

print("2. Female first-class passengers:",
      count_female_first_class)

# Survival percentage of female first-class passengers

print("   Survival percentage:",
      survival_percentage, "%")

# 3. Age 20-40, Fare above median, and survived

print("3. Age 20-40, Fare above median, and survived:",
      count_age_20_40)

# 4. Travelling alone, age < 30, and did not survive

print("4. Travelling alone, age < 30, and did not survive:",
      count_alone_young_not_survived)

# 5. Embarked S, Pclass 2 or 3, Fare above Southampton median

print("5. Embarked S, Pclass 2 or 3, Fare above Southampton median:",
      count_south_pclass_high_fare)
```

##### Task 6 - groupby() and Aggregation


● Survival rate by Sex.
● Survival rate by Pclass.
● Average Age and Fare by Pclass.
● Passenger count and survival rate by Sex-Pclass.
● Passenger count, average Fare, and survival rate by Embarked

```python
# Group the passengers according to their Sex.
#
# Then calculate the mean of the Survived column.
#
# Survived contains:
# 1 -> passenger survived
# 0 -> passenger did not survive
#
# Therefore, the mean of Survived gives the survival rate.

survival_by_sex = df.groupby("Sex")["Survived"].mean()

# Display the survival rate.

print("Survival Rate by Sex:")
print(survival_by_sex)
```

```python
# Group passengers according to their passenger class.
#
# Pclass contains:
# 1 -> First class
# 2 -> Second class
# 3 -> Third class
#
# We calculate the mean of Survived for each class.
# This gives the survival rate for each passenger class.

survival_by_pclass = df.groupby("Pclass")["Survived"].mean()

# Display the survival rate.

print("Survival Rate by Pclass:")
print(survival_by_pclass)
```

```python
# Group the passengers according to Pclass.
#
# Then calculate the average Age and average Fare
# for each passenger class.
#
# agg(["mean"]) is not necessary here.
# We can directly use mean() after selecting the columns.

average_age_fare = df.groupby("Pclass")[["Age", "Fare"]].mean()

# Display the result.

print("Average Age and Fare by Pclass:")
print(average_age_fare)
```

```python
# Group the passengers by BOTH Sex and Pclass.
#
# This creates groups such as:
# female + 1st class
# female + 2nd class
# female + 3rd class
# male + 1st class
# male + 2nd class
# male + 3rd class
#
# For each group:
# count -> counts the number of passengers
# mean  -> calculates the survival rate

sex_pclass_summary = df.groupby(
    ["Sex", "Pclass"]
).agg(
    Passenger_Count=("PassengerId", "count"),
    Survival_Rate=("Survived", "mean")
)

# Display the result.

print("Passenger Count and Survival Rate by Sex-Pclass:")
print(sex_pclass_summary)
```

```python
# Convert the survival rate from decimal to percentage.
#
# For example:
# 0.742038 × 100 = 74.2038%

sex_pclass_summary["Survival_Rate"] = (
    sex_pclass_summary["Survival_Rate"] * 100
)

# Display the updated result.

print("Sex-Pclass Summary:")
print(sex_pclass_summary)
```

```python
# Group passengers according to the port where they embarked.
#
# For each port, we calculate:
#
# Passenger_Count
# → Number of passengers
#
# Average_Fare
# → Average Fare paid by passengers
#
# Survival_Rate
# → Average value of Survived
# → This gives the survival rate

embarked_summary = df.groupby("Embarked").agg(
    Passenger_Count=("PassengerId", "count"),
    Average_Fare=("Fare", "mean"),
    Survival_Rate=("Survived", "mean")
)

# Display the result.

print("Passenger Count, Average Fare, and Survival Rate by Embarked:")
print(embarked_summary)
```

```python
# Convert the survival rate from decimal to percentage.
#
# For example:
# 0.553571 becomes 55.3571%.

embarked_summary["Survival_Rate"] = (
    embarked_summary["Survival_Rate"] * 100
)

# Display the final result.

print("Embarked Summary:")
print(embarked_summary)
```

```python
# Display all the results from Task 6 together.

print("TASK 6 RESULTS")
print("=" * 50)


# 1. Survival rate by Sex

print("\n1. Survival Rate by Sex:")
print(survival_by_sex * 100)


# 2. Survival rate by Pclass

print("\n2. Survival Rate by Pclass:")
print(survival_by_pclass * 100)


# 3. Average Age and Fare by Pclass

print("\n3. Average Age and Fare by Pclass:")
print(average_age_fare)


# 4. Passenger count and survival rate by Sex-Pclass

print("\n4. Passenger Count and Survival Rate by Sex-Pclass:")
print(sex_pclass_summary)


# 5. Passenger count, average Fare, and survival rate by Embarked

print("\n5. Passenger Count, Average Fare, and Survival Rate by Embarked:")
print(embarked_summary)
```

```python
# isnull() checks every cell in the DataFrame.
#
# If a cell contains a missing value (NaN),
# isnull() returns True.
#
# sum() then counts how many True values
# are present in each column.
#
# Therefore, this gives us the number of
# missing values in every column.

missing_count = df.isnull().sum()

# Display the missing-value count for every column.

print("Missing values in each column:")
print(missing_count)
```

```python
# Calculate the percentage of missing values in each column.
#
# df.isnull().sum()
# → gives the number of missing values.
#
# len(df)
# → gives the total number of rows.
#
# Dividing missing values by total rows and
# multiplying by 100 gives the percentage.

missing_percentage = (df.isnull().sum() / len(df)) * 100

# Display the missing-value percentage.

print("Missing value percentage in each column:")
print(missing_percentage)
```

```python
# Count the number of missing values in the Age column
# before filling them.

age_missing_before = df["Age"].isnull().sum()

# Display the count.

print("Missing Age values before filling:",
      age_missing_before)
```

```python
# Calculate the mean Age.
#
# Pandas automatically ignores NaN values
# when calculating the mean.

mean_age = df["Age"].mean()

# Display the mean Age.

print("Mean Age:", mean_age)
```

```python
# Fill all missing Age values with the mean Age.
#
# fillna() is used to replace missing values.
#
# inplace=True means that the change is made
# directly in the DataFrame.

df["Age"].fillna(mean_age, inplace=True)

# Check the number of missing Age values after filling.

age_missing_after = df["Age"].isnull().sum()

# Display the result.

print("Missing Age values after filling:",
      age_missing_after)
```

```python
# Create a copy of the original DataFrame.
# This allows us to try mean imputation without
# changing the original data.

df_mean = df.copy()

# Calculate the mean Age.

mean_age = df_mean["Age"].mean()

# Replace missing Age values with the mean Age.

df_mean["Age"] = df_mean["Age"].fillna(mean_age)

# Count missing Age values after imputation.

print("Missing Age values after mean imputation:",
      df_mean["Age"].isnull().sum())
```

```python
# Create another copy of the DataFrame.

df_median = df.copy()

# Calculate the median Age.
#
# median() finds the middle value of the Age values.

median_age = df_median["Age"].median()

# Replace missing Age values with the median Age.

df_median["Age"] = df_median["Age"].fillna(median_age)

# Count missing Age values after imputation.

print("Missing Age values after median imputation:",
      df_median["Age"].isnull().sum())
```

```python
# Create another copy of the DataFrame.

df_mode = df.copy()

# Calculate the mode of Age.
#
# mode() can return more than one value,
# so [0] selects the first mode.

mode_age = df_mode["Age"].mode()[0]

# Replace missing Age values with the mode.

df_mode["Age"] = df_mode["Age"].fillna(mode_age)

# Count missing Age values after imputation.

print("Missing Age values after mode imputation:",
      df_mode["Age"].isnull().sum())
```

```python
# Create another copy of the DataFrame.

df_random = df.copy()

# Find all Age values that are not missing.
#
# dropna() removes the missing Age values,
# leaving only valid Age values.

valid_ages = df_random["Age"].dropna()

# Count how many Age values are missing.

missing_age_count = df_random["Age"].isnull().sum()

# Select random Age values from the valid Age values.
#
# random_state=42 makes the random selection
# reproducible.
#
# n=missing_age_count means we select exactly
# as many random values as there are missing values.

random_ages = valid_ages.sample(
    n=missing_age_count,
    replace=True,
    random_state=42
).values

# Replace the missing Age values with the
# randomly selected Age values.

df_random.loc[df_random["Age"].isnull(), "Age"] = random_ages

# Count missing Age values after random imputation.

print("Missing Age values after random imputation:",
      df_random["Age"].isnull().sum())
```

```python
# Display the number of missing Age values
# after using each imputation method.

print("Missing Age values after different methods:")
print("--------------------------------------------")

print("Mean imputation:",
      df_mean["Age"].isnull().sum())

print("Median imputation:",
      df_median["Age"].isnull().sum())

print("Mode imputation:",
      df_mode["Age"].isnull().sum())

print("Random value imputation:",
      df_random["Age"].isnull().sum())
```

```python
# Calculate Q1 (first quartile).
#
# Q1 is the 25th percentile of Fare.

Q1 = df["Fare"].quantile(0.25)

# Calculate Q3 (third quartile).
#
# Q3 is the 75th percentile of Fare.

Q3 = df["Fare"].quantile(0.75)

# Display Q1 and Q3.

print("Q1:", Q1)
print("Q3:", Q3)
```

```python
# Calculate the Interquartile Range (IQR).
#
# Formula:
#
# IQR = Q3 - Q1
#
# IQR represents the spread of the middle 50%
# of the Fare values.

IQR = Q3 - Q1

# Display the IQR.

print("IQR:", IQR)
```

```python
# Calculate the lower outlier bound.
#
# Formula:
# Lower Bound = Q1 - 1.5 × IQR

lower_bound = Q1 - (1.5 * IQR)


# Calculate the upper outlier bound.
#
# Formula:
# Upper Bound = Q3 + 1.5 × IQR

upper_bound = Q3 + (1.5 * IQR)


# Display both bounds.

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
# Select Fare values below the lower bound
# OR above the upper bound.
#
# The | symbol means OR.

fare_outliers = df[
    (df["Fare"] < lower_bound) |
    (df["Fare"] > upper_bound)
]

# Count the number of outliers.

outlier_count = len(fare_outliers)

# Display the number of outliers.

print("Number of Fare outliers:", outlier_count)
```

```python
# Display only the Fare values that are identified
# as outliers.

print("Fare outlier values:")

print(fare_outliers["Fare"])
```

```python
# Display all the important Fare outlier calculations
# together in one place.

print("FARE OUTLIER SUMMARY")
print("=" * 40)

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

print("Number of Outliers:", outlier_count)
```

##### Task 8 - Features and Pivot Table


● Create FamilySize = SibSp + Parch + 1 and IsAlone = 1 when FamilySize = 1.
● Create a pivot table: rows=Sex, columns=Pclass, values=mean Survived. Find the highest and lowest groups.

```python
# Create a new column called FamilySize.
#
# SibSp = number of siblings and spouses travelling
#         with the passenger.
#
# Parch = number of parents and children travelling
#         with the passenger.
#
# We add 1 because the passenger themselves
# is also a member of the family.
#
# Formula:
# FamilySize = SibSp + Parch + 1

df["FamilySize"] = df["SibSp"] + df["Parch"] + 1

# Display the first 5 rows to check the new column.

print(df[["SibSp", "Parch", "FamilySize"]].head())
```

```python
# Create a new column called IsAlone.
#
# If FamilySize = 1:
# → the passenger is travelling alone
# → assign 1
#
# If FamilySize is greater than 1:
# → the passenger is travelling with family
# → assign 0
#
# The condition (df["FamilySize"] == 1) gives
# True or False values.
#
# astype(int) converts:
# True  -> 1
# False -> 0

df["IsAlone"] = (df["FamilySize"] == 1).astype(int)

# Display the first 10 rows to check the new columns.

print(df[["SibSp", "Parch", "FamilySize", "IsAlone"]].head(10))
```

```python
# Display the important columns together.
#
# This allows us to verify that FamilySize and IsAlone
# have been created correctly.

print(df[[
    "SibSp",
    "Parch",
    "FamilySize",
    "IsAlone"
]].head(10))
```

```python
# Create a pivot table using pd.pivot_table().
#
# index="Sex"
# → Sex will be used as the rows.
#
# columns="Pclass"
# → Pclass will be used as the columns.
#
# values="Survived"
# → We want to calculate values from the Survived column.
#
# aggfunc="mean"
# → Calculate the mean survival value.
#
# Since Survived contains:
# 1 = survived
# 0 = did not survive
#
# the mean represents the survival rate.

pivot_table = pd.pivot_table(
    df,
    index="Sex",
    columns="Pclass",
    values="Survived",
    aggfunc="mean"
)

# Display the pivot table.

print("Survival Rate Pivot Table:")
print(pivot_table)
```

```python
# Convert the survival rates from decimal values
# into percentages.

pivot_percentage = pivot_table * 100

# Display the pivot table as percentages.

print("Survival Rate Pivot Table (%):")
print(pivot_percentage)
```

```python
# Find the position of the highest value
# in the pivot table.
#
# idxmax() returns the row and column labels
# where the maximum value occurs.

highest_group = pivot_table.stack().idxmax()

# Find the actual highest survival rate.

highest_rate = pivot_table.stack().max()

# Display the result.

print("Highest survival group:")
print("Sex:", highest_group[0])
print("Pclass:", highest_group[1])
print("Survival Rate:", highest_rate * 100, "%")
```

```python
# Find the position of the lowest survival rate
# in the pivot table.

lowest_group = pivot_table.stack().idxmin()

# Find the actual lowest survival rate.

lowest_rate = pivot_table.stack().min()

# Display the result.

print("Lowest survival group:")
print("Sex:", lowest_group[0])
print("Pclass:", lowest_group[1])
print("Survival Rate:", lowest_rate * 100, "%")
```

```python
# Display the final results of Task 8 together.

print("TASK 8 RESULTS")
print("=" * 50)


# Display the first few rows with the new features.

print("\n1. New Features:")
print(df[[
    "SibSp",
    "Parch",
    "FamilySize",
    "IsAlone"
]].head())


# Display the pivot table.

print("\n2. Survival Rate Pivot Table:")
print(pivot_percentage)


# Display the highest survival group.

print("\n3. Highest Survival Group:")
print("Sex:", highest_group[0])
print("Pclass:", highest_group[1])
print("Survival Rate:", highest_rate * 100, "%")


# Display the lowest survival group.

print("\n4. Lowest Survival Group:")
print("Sex:", lowest_group[0])
print("Pclass:", lowest_group[1])
print("Survival Rate:", lowest_rate * 100, "%")
```

● Plot a correlation heatmap for relevant numerical columns. Note the strongest positive and negative relationships.

```python
# Import Matplotlib for creating plots.

import matplotlib.pyplot as plt

# Import Seaborn for creating the correlation heatmap
# and other statistical visualizations.

import seaborn as sns
```

```python
# Select the numerical columns that are useful for analysis.
#
# PassengerId is not included because it is only an
# identification number and does not represent a
# meaningful numerical feature.
#
# Survived -> whether passenger survived
# Pclass   -> passenger class
# Age      -> passenger age
# SibSp    -> siblings/spouses
# Parch    -> parents/children
# Fare     -> ticket fare

numerical_columns = [
    "Survived",
    "Pclass",
    "Age",
    "SibSp",
    "Parch",
    "Fare"
]

# Create a DataFrame containing only these columns.

numerical_data = df[numerical_columns]

# Display the selected numerical data.

print(numerical_data.head())
```

```python
# Calculate the correlation between all selected
# numerical columns.
#
# corr() calculates the Pearson correlation
# between the numerical variables.

correlation = numerical_data.corr()

# Display the correlation values.

print("Correlation Matrix:")
print(correlation)
```

```python
# Create the figure size.
# figsize controls the width and height of the graph.

plt.figure(figsize=(8, 6))

# Create the correlation heatmap.
#
# correlation
# → the correlation matrix we calculated above
#
# annot=True
# → displays the actual correlation values inside
#   the heatmap cells
#
# fmt=".2f"
# → displays the values up to 2 decimal places

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f"
)

# Add a title to the heatmap.

plt.title("Correlation Heatmap")

# Display the heatmap.

plt.show()
```

```python
# Make a copy of the correlation matrix.

corr_copy = correlation.copy()

# Set the diagonal values to NaN.
#
# The diagonal contains 1.00 because every variable
# has a perfect correlation with itself.
#
# We don't want to consider these values when finding
# the strongest relationships.

import numpy as np

np.fill_diagonal(corr_copy.values, np.nan)

# Find the strongest positive correlation.

strongest_positive = corr_copy.stack().idxmax()
positive_value = corr_copy.stack().max()

# Find the strongest negative correlation.

strongest_negative = corr_copy.stack().idxmin()
negative_value = corr_copy.stack().min()

# Display the results.

print("Strongest Positive Relationship:")
print(strongest_positive, "=", positive_value)

print("\nStrongest Negative Relationship:")
print(strongest_negative, "=", negative_value)
```

```python
# Group the passengers by Sex.
#
# Then calculate the mean of Survived.
#
# Since:
# 1 = survived
# 0 = did not survive
#
# the mean gives the survival rate.

survival_by_sex = df.groupby("Sex")["Survived"].mean()

# Convert the survival rate from decimal to percentage.

survival_by_sex_percentage = survival_by_sex * 100

# Display the survival rates.

print("Survival Rate by Sex:")
print(survival_by_sex_percentage)
```

```python
# Create a bar chart showing the survival rate
# for male and female passengers.

plt.figure(figsize=(6, 5))

# Create the bar chart.
#
# survival_by_sex_percentage.index
# → male and female
#
# survival_by_sex_percentage.values
# → their survival percentages
#
# edgecolor="black"
# → adds a border around each bar.

plt.bar(
    survival_by_sex_percentage.index,
    survival_by_sex_percentage.values,
    edgecolor="black"
)

# Add a title.

plt.title("Survival Rate by Sex")

# Add a label to the x-axis.

plt.xlabel("Sex")

# Add a label to the y-axis.

plt.ylabel("Survival Rate (%)")

# Set the y-axis from 0 to 100
# because survival rate is a percentage.

plt.ylim(0, 100)

# Display the graph.

plt.show()
```
