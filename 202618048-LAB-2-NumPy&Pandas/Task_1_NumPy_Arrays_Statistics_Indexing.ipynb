```python
# Import NumPy.
# NumPy is used for working with arrays and performing numerical operations.
import numpy as np
```

```python
# Set the random seed so that the same random numbers are generated every time we run the code.
# This makes our result reproducible.
np.random.seed(42)

# Generate Array A containing exactly 100 random integers.
# randint(1, 101, 100) generates 100 integers from 1 to 100.
A = np.random.randint(1, 101, 100)

# Display Array A.
print("Array A:")
print(A)
```

```python
# Find the minimum value in Array A.
print("Minimum:", np.min(A))

# Find the maximum value in Array A.
print("Maximum:", np.max(A))

# Find the median value of Array A.
print("Median:", np.median(A))

# Find the mean (average) value of Array A.
print("Mean:", np.mean(A))

# Find the standard deviation of Array A.
print("Standard Deviation:", np.std(A))
```

```python
# Generate Array B with exactly 100 values using np.arange().
# arange(0, 100) starts at 0 and stops before 100.
# Therefore, it gives the values 0 to 99, which are exactly 100 values.
B = np.arange(0, 100)

# Display Array B.
print("Array B:")
print(B)

# Check the number of values in Array B.
print("Number of values:", len(B))
```

```python
# np.zeros() creates an array filled with zeros.
# Here we create 5 zeros.
zeros_array = np.zeros(5)

# np.ones() creates an array filled with ones.
# Here we create 5 ones.
ones_array = np.ones(5)

print("Zeros array:", zeros_array)
print("Zeros shape:", zeros_array.shape)
print("Zeros data type:", zeros_array.dtype)

print("Ones array:", ones_array)
print("Ones shape:", ones_array.shape)
print("Ones data type:", ones_array.dtype)
```

```python
# np.linspace() creates a specified number of equally spaced values.
# Here, 5 values are created between 0 and 10, including both 0 and 10.
linear_array = np.linspace(0, 10, 5)

print("Array created using np.linspace():")
print(linear_array)

# Difference:
# np.arange() mainly uses a step size, while np.linspace() mainly uses the number of values.
# Example: np.arange(0, 10, 2) -> 0, 2, 4, 6, 8
# Example: np.linspace(0, 10, 5) -> 0, 2.5, 5, 7.5, 10
```

```python
# Create a simple 2D array with 3 rows and 4 columns.
# The array contains numbers from 1 to 12.
array_2d = np.arange(1, 13).reshape(3, 4)

print("2D Array:")
print(array_2d)

# shape tells us the number of rows and columns.
print("Shape:", array_2d.shape)

# ndim tells us the number of dimensions.
print("Number of dimensions:", array_2d.ndim)

# Indexing: Python indexing starts from 0.
# [0, 0] means first row and first column.
print("First element:", array_2d[0, 0])

# Select the first row.
print("First row:", array_2d[0, :])

# Select the second column.
print("Second column:", array_2d[:, 1])

# Slicing: select the first two rows and first three columns.
print("Sliced part:")
print(array_2d[0:2, 0:3])
```

```python
# Create a simple 3D array.
# 2 blocks, each block has 3 rows and 4 columns.
# Therefore, its shape is (2, 3, 4).
array_3d = np.arange(1, 25).reshape(2, 3, 4)

print("3D Array:")
print(array_3d)

# Show the shape of the 3D array.
print("Shape:", array_3d.shape)

# Show the number of dimensions.
print("Number of dimensions:", array_3d.ndim)

# Indexing: [0, 0, 0] means first block, first row, first column.
print("First element:", array_3d[0, 0, 0])

# Select the first block.
print("First block:")
print(array_3d[0])

# Select the first row from the first block.
print("First row of first block:", array_3d[0, 0, :])

# Select the second column from the first block.
print("Second column of first block:", array_3d[0, :, 1])

# Slicing: select both blocks, first two rows, and first two columns.
print("Sliced part:")
print(array_3d[:, 0:2, 0:2])
```

```python
# Start with a 1D array containing 20 values.
original_array = np.arange(1, 21)

# reshape() changes the shape of an array without changing its values.
# Here, 20 values are changed into a matrix with 4 rows and 5 columns.
matrix = original_array.reshape(4, 5)

print("Matrix after reshape():")
print(matrix)

# flatten() converts the matrix back into a single 1D array.
flattened_array = matrix.flatten()

print("Array after flatten():")
print(flattened_array)
```
