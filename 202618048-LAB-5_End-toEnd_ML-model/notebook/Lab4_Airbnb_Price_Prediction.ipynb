##### 4.4 Interquartile Range (IQR) Analysis of Price

The Interquartile Range (IQR) method is used to statistically identify potential price outliers.

The IQR method is based on the first quartile (Q1) and third quartile (Q3).

The potential outlier boundaries are calculated as:

Lower Bound = Q1 - 1.5 × IQR

Upper Bound = Q3 + 1.5 × IQR

Values outside these boundaries are considered statistical outliers.

However, potential outliers will not be removed automatically. The number and characteristics of these listings will be examined before deciding whether removal is appropriate.

```python
# ============================================================
# CELL 24: CALCULATING IQR OUTLIER BOUNDARIES FOR PRICE
# ============================================================

# Calculating the first quartile (25th percentile).

Q1 = df_working["price"].quantile(0.25)


# Calculating the third quartile (75th percentile).

Q3 = df_working["price"].quantile(0.75)


# Calculating the Interquartile Range.
#
# Formula:
#
# IQR = Q3 - Q1

IQR = Q3 - Q1


# Calculating the lower outlier boundary.
#
# Formula:
#
# Lower Bound = Q1 - 1.5 * IQR

lower_bound = Q1 - (1.5 * IQR)


# Calculating the upper outlier boundary.
#
# Formula:
#
# Upper Bound = Q3 + 1.5 * IQR

upper_bound = Q3 + (1.5 * IQR)


# Displaying all important values.

print("First Quartile (Q1):", Q1)

print("Third Quartile (Q3):", Q3)

print("Interquartile Range (IQR):", IQR)

print("Lower Outlier Boundary:", lower_bound)

print("Upper Outlier Boundary:", upper_bound)
```

```python
# ============================================================
# CELL 25: COUNTING POTENTIAL PRICE OUTLIERS
# ============================================================

# Identifying prices below the lower boundary.

lower_outliers = df_working[
    df_working["price"] < lower_bound
]


# Identifying prices above the upper boundary.

upper_outliers = df_working[
    df_working["price"] > upper_bound
]


# Calculating the total number of potential outliers.

total_price_outliers = (
    len(lower_outliers)
    +
    len(upper_outliers)
)


# Calculating the percentage of potential outliers.

price_outlier_percentage = (
    total_price_outliers
    /
    len(df_working)
) * 100


# Displaying the results.

print(
    "Number of potential lower price outliers:",
    len(lower_outliers)
)

print(
    "Number of potential upper price outliers:",
    len(upper_outliers)
)

print(
    "Total potential price outliers:",
    total_price_outliers
)

print(
    "Percentage of potential price outliers:",
    round(price_outlier_percentage, 2),
    "%"
)
```

##### 4.5 Detailed Investigation of Extreme Price Values

However, Airbnb listings above the IQR upper boundary may still represent legitimate high-value or luxury properties.

Therefore, before removing any records, the upper tail of the price distribution is examined using percentiles.

Percentiles help identify the difference between moderately high prices and extremely rare price values.

```python
# ============================================================
# CELL 26: EXAMINING IMPORTANT PRICE PERCENTILES
# ============================================================

# We calculate multiple percentiles for the 'price' column.
#
# These values help us understand how prices are distributed
# in the upper part of the dataset.
#
# For example:
#
# 90th percentile:
# 90% of listings have a price less than or equal to this value.
#
# 99th percentile:
# 99% of listings have a price less than or equal to this value.
#
# The higher percentiles are particularly useful for
# investigating extremely expensive listings.

price_percentiles = df_working["price"].quantile(
    [
        0.50,   # Median
        0.75,   # 75th percentile
        0.90,   # 90th percentile
        0.95,   # 95th percentile
        0.99,   # 99th percentile
        0.995,  # 99.5th percentile
        0.999   # 99.9th percentile
    ]
)


# Displaying the percentile values.

print("Important Price Percentiles:\n")

print(price_percentiles)
```

##### 4.6 High-Price Listing Analysis

The number of Airbnb listings above different high-price thresholds is calculated.

This analysis helps determine whether extremely high prices represent a large portion of the dataset or only a very small and unusual group of observations.

The results will be used together with percentile analysis before deciding on an appropriate extreme outlier treatment strategy.

```python
# ============================================================
# CELL 28: COUNTING LISTINGS ABOVE HIGH-PRICE THRESHOLDS
# ============================================================

# Defining several high-price thresholds for investigation.
#
# These thresholds are used only for analysis at this stage.
# No rows are removed here.

price_thresholds = [
    500,
    750,
    1000,
    1500,
    2000,
    3000,
    5000
]


# Creating an empty list to store the results.

threshold_results = []


# Looping through each price threshold.

for threshold in price_thresholds:
    
    # Counting the number of listings with a price
    # greater than the current threshold.
    
    count = (
        df_working["price"] > threshold
    ).sum()
    
    
    # Calculating the percentage of the complete dataset
    # represented by these listings.
    
    percentage = (
        count / len(df_working)
    ) * 100
    
    
    # Adding the results to our list.
    
    threshold_results.append(
        {
            "Price Greater Than": threshold,
            "Number of Listings": count,
            "Percentage of Dataset": round(
                percentage,
                3
            )
        }
    )


# Converting the results into a DataFrame.

high_price_summary = pd.DataFrame(
    threshold_results
)


# Displaying the summary table.

high_price_summary
```

```python
# ============================================================
# CELL 29: DISPLAYING THE MOST EXPENSIVE AIRBNB LISTINGS
# ============================================================

# Selecting important columns that will help us understand
# the most expensive listings.
#
# We do not need to display every column because the complete
# dataset contains many features.

most_expensive_listings = df_working[
    [
        "id",
        "name",
        "neighbourhood_group",
        "neighbourhood",
        "room_type",
        "price",
        "minimum_nights",
        "number_of_reviews",
        "availability_365"
    ]
].sort_values(
    
    # Sorting listings from highest price to lowest price.
    
    by="price",
    
    ascending=False
)


# Displaying the 20 most expensive listings.

most_expensive_listings.head(20)
```
